```python
particles_dict = {}

for i, event in enumerate(events):
    # take only final state particles
    for p in event.particles:
        if p.status != 1: 
            continue
        if p.pid not in particles_dict:
            particles_dict[p.pid] = {}
        
```

```python
def is_inside_box(x,y,z):
    # FASER dimensions (in mm)
    lx = 2.5e2 # 25 cm
    ly = 3e2 # 30 cm 
    lz = 1e3 # 1 m
    if ((x <= lx/2) and (x >= -lx/2)) and ((y <= ly/2) and (y >= -ly/2)) and ((z <= lz/2) and (z >= -lz/2)):
        return True
    else:
        return False
```

```python
# position four-vector is given in mm. To obtain t in seconds, divide by c, where c = 2.998*e11 mm/s
proton_id = 2212
piplus_id = 211
piminus_id = -211
photon_id = 22
ngamma = 0
ngamma_inbox = 0
npiplus = 0
npiplus_inbox = 0
nproton = 0
nproton_inbox = 0
n = 0
npart = 0
nprod = 0


for i, event in enumerate(events):

    # print("\n---------\nEvent %i\n----------\n"%(i))
    for p in event.particles:
        if p.status != 1:  # not "final state"
            continue
        npart += 1

        if p.production_vertex:
            nprod += 1
    
        # if p.pid == proton_id:
        #     print("Proton")
        #     if p.production_vertex:
        #         print("produced at", p.production_vertex.position)
        #         for parent in p.production_vertex.particles_in:
        #             print("parents:",parent.pid, parent.status,"\n")

        # if p.pid == piplus_id:
        #     print("Pion plus")
        #     if p.production_vertex:
        #         print("produced at", p.production_vertex.position)
        #         for parent in p.production_vertex.particles_in:
        #             print("(parent id, status):",parent.pid, parent.status,p.momentum.e,"\n")
        
        if p.pid == photon_id:
            ngamma += 1
            # print("Photon")
            if p.production_vertex:
                n += 1
                # print("produced at", p.production_vertex.position)
                # for parent in p.production_vertex.particles_in:
                    # print("(parent id, status):",parent.pid, parent.status,p.momentum.e,"\n")
                v = p.production_vertex.position
                if is_inside_box(v.x, v.y, v.z):
                    ngamma_inbox += 1
                    
```
